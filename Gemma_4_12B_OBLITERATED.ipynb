```python

import os
import platform
import shutil

print("CPU:", platform.processor() or platform.machine())
print("CPU cores:", os.cpu_count())

with open("/proc/meminfo") as f:
    for line in f:
        if line.startswith(("MemTotal:", "MemAvailable:", "SwapTotal:", "SwapFree:")):
            print(line.strip())

print("\nDisk:")
print(shutil.disk_usage("/"))
print("\nHome directory:", os.path.expanduser("~"))
```

```python
import os

roots = ["/root", "/content", "/mnt", "/tmp"]
extensions = (".safetensors", ".gguf", ".bin")

for root in roots:
    if not os.path.exists(root):
        continue

    print(f"\nSearching {root}...")
    for current, dirs, files in os.walk(root):
        dirs[:] = [d for d in dirs if d not in {
            ".git", "node_modules", "__pycache__"
        }]

        for filename in files:
            if filename.lower().endswith(extensions):
                path = os.path.join(current, filename)
                try:
                    size = os.path.getsize(path) / (1024**3)
                    if size >= 0.1:
                        print(f"{size:.2f} GiB  {path}")
                except OSError:
                    pass
```

```python

import os

paths = [
    "/root/.cache/huggingface",
    "/home",
    "/workspace",
    "/opt",
    "/kaggle/working",
    "/kaggle/input",
]

for path in paths:
    if os.path.exists(path):
        print(f"\nFOUND: {path}")
        for current, dirs, files in os.walk(path):
            dirs[:] = [d for d in dirs if d not in {".git", "__pycache__", "node_modules"}]
            for filename in files:
                if filename.endswith((".safetensors", ".gguf", ".bin")):
                    full_path = os.path.join(current, filename)
                    size_gb = os.path.getsize(full_path) / (1024**3)
                    if size_gb >= 0.1:
                        print(f"{size_gb:.2f} GiB  {full_path}")
    else:
        print(f"NOT FOUND: {path}")
```

```python
# !pip install llama-cpp-python

from llama_cpp import Llama

llm = Llama.from_pretrained(
	repo_id="OBLITERATUS/Gemma-4-12B-OBLITERATED",
	filename="Gemma-4-12B-OBLITERATED-BF16.gguf",
)
```

```python
llm.create_chat_completion(
	messages = [
		{
			"role": "user",
			"content": "What is the capital of France?"
		}
	]
)
```
